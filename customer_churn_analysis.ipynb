```python
# Step 1 : Import Liberaries

import pandas as pd
import numpy as np
```

```python
# Step 2 : Load Dataset

df = pd.read_excel("C:/Users/shyam/Downloads/Python/Portfolio Projects/Customer Churn DL/customer_churn_data.xlsx")
```

```python
# Remove Duplicates

df.drop_duplicates(inplace = True)
```

```python
df.shape
```

```python
# Step 6 : Replace Dirty Values

df.replace(['N/A','NULL','',' '],np.nan,inplace = True)
```

```python
df.isnull().sum()
```

```python
# Step 7 : Remove Extra Spaces

df = df.apply(lambda x:x.str.strip() if x.dtype == 'object' else x)
```

```python
# value counts

df["Subscription_Type"].value_counts()
```

```python
df.dtypes
```

```python
# Step 8 : Proper Case

df["Customer_Name"] = df["Customer_Name"].str.title()
df["Gender"] = df["Gender"].str.title()
df["State"] = df["State"].str.title()
df["City"] = df["City"].str.title()
df["Subscription_Type"] = df["Subscription_Type"].str.title()
df["Contract_Type"] = df["Contract_Type"].str.title()
df["Payment_Method"] = df["Payment_Method"].str.title()
df["Internet_Service"] = df["Internet_Service"].str.title()
df["Tech_Support"] = df["Tech_Support"].str.title()
df["Senior_Citizen"] = df["Senior_Citizen"].str.title()
df["Dependents"] = df["Dependents"].str.title()
df["Churn"] = df["Churn"].str.title()

```

```python
df["Subscription_Type"].value_counts()
```

```python
# Step 9 : Convert Numeric Columns

cols = ["Age","Tenure_Months","Monthly_Charges","Total_Charges"]

for x in cols :
    df[x] = pd.to_numeric(df[x],errors ="coerce" )
```

```python
df.describe()
```

```python
# Step 10 : Remove Invalid Age

df = df[(df["Age"]>=18) & (df["Age"]<=100)]
```

```python
# Step 11 : Remove Negative Charges

df = df[df["Monthly_Charges"]>=0]
df = df[df["Total_Charges"]>=0]
```

```python
df.shape
```

```python
# Step 12 : Convert Date

df["Last_Interaction_Date"] = pd.to_datetime (df["Last_Interaction_Date"],errors ="coerce")
```

```python
# Step 13 : Fill Missing Value

df["Tech_Support"].fillna("No",inplace = True)
df["Payment_Method"].fillna("Unknown",inplace =True)
df["Monthly_Charges"].fillna(df["Monthly_Charges"].mean(),inplace =True)
df["Tenure_Months"].fillna(df["Tenure_Months"].mean(),inplace =True)
df["Age"].fillna(df["Age"].mean(),inplace =True)
```

```python
df["Last_Interaction_Date"].fillna(df["Last_Interaction_Date"].median(),inplace =True)
```

```python
# Step 14 : Feature Engineering

# Customer Lifetime Value

df["Customer_Value"] = df["Monthly_Charges"] * df["Tenure_Months"]
```

```python
df.head()
```

```python
# Monthly Revenue

df["Monthly_Revenue"] = df["Monthly_Charges"]
```

```python
# Tenure Group "0-12","13-24","25-48","49-72"

bins = [0,12,24,48,72]
labels = ["0-12 month","13-24 month","25-48 month","49-72 month"]
df["Tenure_Group"] = pd.cut(df["Tenure_Months"],bins = bins, labels = labels)
```

```python
df.head()
```

```python
# Senior Citizen Flag

df["Senior_Flag"] = np.where(df["Age"]>=60,"Senior","Adult")
```

```python
df.head()
```

```python
# Churn Flag

df["Churn_Flag"] = df["Churn"].map({"Yes":1,"No":0})
```
